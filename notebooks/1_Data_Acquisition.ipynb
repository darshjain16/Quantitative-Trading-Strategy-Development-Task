# Notebook 1: Data Acquisition

```python
import pandas as pd
import numpy as np
import os

print(" Starting Data Acquisition...")

# 1. Generate Synthetic Spot Data (Simulating Nifty 50)
# We create 1 year of 5-min data
dates = pd.date_range(start="2015-01-01", end="2016-01-01", freq="5min")
# Filter for market hours (9:15 to 15:30)
market_dates = [d for d in dates if d.time() >= pd.to_datetime("09:15").time() and d.time() <= pd.to_datetime("15:30").time()]

df_spot = pd.DataFrame(index=market_dates)
df_spot['open'] = 100 + np.cumsum(np.random.randn(len(market_dates)))
df_spot['high'] = df_spot['open'] + np.abs(np.random.randn(len(market_dates)))
df_spot['low'] = df_spot['open'] - np.abs(np.random.randn(len(market_dates)))
df_spot['close'] = df_spot['open'] + np.random.randn(len(market_dates)) * 0.5
df_spot['volume'] = np.random.randint(1000, 10000, len(market_dates))

# 2. Generate Synthetic Futures Data
# Futures usually trade at a premium/discount (Basis)
df_fut = df_spot.copy()
df_fut['close'] = df_fut['close'] * (1 + 0.005 * np.random.randn(len(df_fut))) # 0.5% basis noise
df_fut['oi'] = np.random.randint(5000000, 10000000, len(df_fut)) # Open Interest

# 3. Generate Synthetic Options Data
# Simulating ATM Strike, IV, and OI
df_opt = pd.DataFrame(index=df_spot.index)
df_opt['atm_strike'] = round(df_spot['close'] / 50) * 50
df_opt['call_iv'] = 15 + np.random.normal(0, 2, size=len(df_spot)) # Avg IV ~15%
df_opt['put_iv'] = 16 + np.random.normal(0, 2, size=len(df_spot)) # Avg IV ~16%
df_opt['call_oi'] = np.random.randint(100000, 500000, size=len(df_spot))
df_opt['put_oi'] = np.random.randint(100000, 500000, size=len(df_spot))

# 4. Save to CSV (Simulating "Raw Downloads")
# Saving with 'date' and 'time' columns to match common API formats
df_spot['date'] = df_spot.index.date
df_spot['time'] = df_spot.index.time
df_fut['date'] = df_fut.index.date
df_fut['time'] = df_fut.index.time
df_opt['date'] = df_opt.index.date
df_opt['time'] = df_opt.index.time

# Ensure 'data' folder exists
if not os.path.exists('../data'):
    os.makedirs('../data')

df_spot.to_csv("../data/nifty_spot_5min.csv")
df_fut.to_csv("../data/nifty_futures_5min.csv")
df_opt.to_csv("../data/nifty_options_5min.csv")

print(" Data Generated & Saved to '../data/' folder.")
```
